```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x, y):
    z = x + 1j*y
    return 6*np.sin(z**2)

def g(x, y):
    z = x + 1j*y
    return 5*z**3
```

```python
phi_arr = np.linspace(0, 2*np.pi, 100)

z_arr = np.exp(1j*phi_arr)
x_arr = np.real(z_arr)
y_arr = np.imag(z_arr)

f_arr = f(x_arr, y_arr)
g_arr = g(x_arr, y_arr)

plt.plot(phi_arr, np.abs(f_arr), label='f')
plt.plot(phi_arr, np.abs(g_arr), label='g')
plt.plot(phi_arr, 6*np.full_like(phi_arr, np.sin(1)), label='|6|')
plt.legend()
plt.show()
```
